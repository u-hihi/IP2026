```python
from google.colab import drive
import cv2
from google.colab.patches import cv2_imshow
import numpy as np
import matplotlib.pyplot as plt
drive.mount('/content/gdrive')
```

```python
img = cv2.imread('/content/gdrive/My Drive/ColabDIP/images/Images/lena256.jpg')
RGB_img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

gimg = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
print(gimg.shape)
cv2_imshow(gimg)
```

```python
meanimg = cv2.blur(gimg,(5,5))
cv2_imshow(meanimg)
```

```python
edgeimg = cv2.addWeighted(gimg, 1, meanimg, -1, 0)
cv2_imshow(edgeimg)
```

```python
unsharpimg = cv2.addWeighted(gimg, 1, edgeimg, 0.2, 0)
cv2_imshow(unsharpimg)
```
