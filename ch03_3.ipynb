```python
import cv2
from google.colab.patches import cv2_imshow
import numpy as np
import matplotlib.pyplot as plt
from google.colab import drive
drive.mount('/content/gdrive')
```

```python
def saturation(value):
  if(value > 255):
    value = 255
  elif(value < 0):
    value = 0
  return value
```

```python
img = cv2.imread('/content/gdrive/My Drive/ColabDIP/images/Images/aero2.bmp')
gimg = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

cv2_imshow(gimg)
```

```python
gOut = np.zeros((gimg.shape[0], gimg.shape[1]),dtype=np.ubyte)
```

```python
min = 256
max = 0

for h in range(gimg.shape[0]):
  for w in range(gimg.shape[1]):
    if gimg[h,w] < min:
      min = gimg[h,w]
    if gimg[h,w] > max:
      max = gimg[h,w]


print(min)
print(max)
```

```python
gMinus = np.zeros((gimg1.shape[0], gimg1.shape[1]),dtype=np.ubyte)
```

```python
for h in range(gimg1.shape[0]):
  for w in range(gimg1.shape[1]):
    val = np.abs(np.int32(gimg1[h,w]) - np.int32(gimg2[h,w])) * 3
    # The 'if val < 0:' check is redundant because np.abs always returns a non-negative value.
    # if val < 0:
    #   val = 0
    gMinus[h,w] = saturation(val)

cv2_imshow(gMinus)
```
