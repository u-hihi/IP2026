```python
img = cv2.imread('/content/gdrive/My Drive/ColabDIP/images/Images/rice.bmp')
RGB_img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

gimg = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
print(gimg.shape)
cv2_imshow(gimg)
```

```python
gBin = np.zeros((gimg.shape[0],gimg.shape[1]))
```

```python
threshold = 128

for i in range(gimg.shape[0]):
  for j in range(gimg.shape[1]):
    if gimg[i,j] > threshold:
      gBin[i,j] = 255

cv2_imshow(gBin)
```

```python
kernel = np.ones((3,3),np.uint8)
eroded = cv2.erode(gBin,kernel,iterations = 1)
cv2_imshow(eroded)
```

```python
kernel = np.ones((3,3),np.uint8)
dilated = cv2.dilate(eroded,kernel,iterations = 1)
cv2_imshow(dilated)
```

```python
#  이 과정을 한 번에 하려면 openimg
kernel = np.ones((3,3),np.uint8)
openimg = cv2.morphologyEx(gBin.astype(np.uint8), cv2.MORPH_OPEN, kernel, iterations = 1)
cv2_imshow(openimg)
```
