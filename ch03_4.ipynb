```python
def saturation(value):
  if(value > 255):
    value = 255
  elif(value < 0):
    value = 0
  return value
```

```python
img = cv2.imread('/content/gdrive/My Drive/ColabDIP/images/Images/lena256.jpg')
gimg = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

cv2_imshow(gimg)
```

```python
gOut = np.zeros((gimg.shape[0], gimg.shape[1]),dtype=np.ubyte)
planeNumber = 1     # 몇 번째 비트 플래인
planeValue = 1    #최상위 비트

for i in range(planeNumber-1):
  planeValue = planeValue * 2
```

```python
for h in range(gimg.shape[0]):
  for w in range(gimg.shape[1]):
    imVal = np.int32(gimg[h,w] / planeValue);
    if(imVal % 2 == 0):
      gOut[h,w] = 0
    else:
      gOut[h,w] = 255

cv2_imshow(gOut)
```
