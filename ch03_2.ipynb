```python
import cv2
from google.colab.patches import cv2_imshow
import numpy as np
import matplotlib.pyplot as plt
from google.colab import drive
drive.mount('/content/gdrive')
```

```python
img = cv2.imread('/content/gdrive/My Drive/ColabDIP/images/Images/girl.jpg')
gimg = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

cv2_imshow(gimg)
```

```python
gPlus = np.zeros((gimg.shape[0], gimg.shape[1]), dtype=np.ubyte)
```

```python
val = 120

for h in range(gimg.shape[0]):
  for w in range(gimg.shape[1]):
    plusVal = np.int32(gimg[h][w]) +val
    if(plusVal > 255):
      plusVal = 255
      gPlus[h][w] = plusVal

cv2_imshow(gPlus)
```

```python
gOrigin = np.zeros((gimg.shape[0], gimg.shape[1]), dtype=np.ubyte)
```

```python
val = 120

for h in range(gimg.shape[0]):
  for w in range(gimg.shape[1]):
    plusVal = np.int32(gPlus[h][w]) - val
    if(plusVal > 255):
      plusVal = 255
      gOrigin[h][w] = plusVal

cv2_imshow(gOrigin)
```
