```python
img = cv2.imread('/content/gdrive/My Drive/ColabDIP/images/Images/cell.bmp')
gimg1 = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
RGB_img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

cv2_imshow(gimg1)
```

```python
hist = cv2.calcHist([gimg1], [0], None, [256], [0, 255])
#hist = cv2.calcHist([RGB_img], [0], None, [256], [0, 255])
plt.plot(hist)
plt.show()
```

```python
gBin = np.zeros((gimg1.shape[0], gimg1.shape[1]),np.uint8)
```

```python
T = 180

for i in range(gimg1.shape[0]):
  for j in range(gimg1.shape[1]):
    if gimg1[i, j] > T:
      gBin[i, j] = 255
    else:
      gBin[i, j] = 0

cv2_imshow(gBin)
```
